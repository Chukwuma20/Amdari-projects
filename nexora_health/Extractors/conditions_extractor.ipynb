```python
import numpy as np
import pandas as pd
from pathlib import Path #this imports the tool for managing file paths/locations
```

```python
from process_specification import dataset_specs #import the dicitionary and dicitonary variable(dataset_specs) from the process_specification.py file
```

```python
data_folder = "csv.files"

def extract(dataset_specs, data_folder):  #Create a function called extract. 
    file_path = Path(data_folder) / dataset_specs["file_name"]  #Create a variable called file_path that combines the data_folder and the file_name from the dataset_specs dictionary.                                      

    df = pd.read_csv(                   #  Read the CSV into a DataFrame.
        file_path,
        dtype=str,
        keep_default_na=False 
    )

    mapping = dataset_specs["column_mapping"]

#when you read a csv into pandas (e.g pd.read_csv ("file.csv")), the column headers on your csv file becomes df.columns.
    actual_columns = set(df.columns)  #Create a variable called actual_columns that contains the set of column names in the DataFrame.

    expected_columns = set(mapping.keys())
    missing_columns = expected_columns - actual_columns

# The specification tells us what columns we expect. We compare those against the CSV. If anything is missing, we stop immediately because the input data doesn't meet our contract.
# That's basically what "fail fast" and "broken batch contract" mean in this task. so this is the fail fast part
    if missing_columns:
          raise ValueError(
            f"Missing expected columns in {dataset_specs['file_name']}: "
            f"{sorted(missing_columns)}"  #sort the results of the missing_columns in alphabetical order 
        )

    df = df[list(mapping.keys())]  #Selects only the columns defined in the specification. df[] filters the DataFrame to include only the columns specified in the mapping dictionary. list(mapping.keys())/ list(mapping) creates a list of the expected column names based on the keys of the mapping dictionary.
    df = df.rename(columns=mapping)
   
    return df
```

```python
conditions_df = extract(dataset_specs["conditions"], data_folder) #call the extract function and pass the conditions dataset specification and the data folder path. The result is stored in conditions_df.
print(conditions_df.columns.tolist())
conditions_df.head()
```
